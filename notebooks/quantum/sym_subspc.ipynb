## Probability Theory

What the formal definition of random variable is requires more measure theory and is beyond the scope of this paper. However, for our purposes, a random variable $X$ is simply a mapping from some sample space $\Omega$ into the space in which we represent it mathematically, in our case $\mathbb{C}^n$. There is nothing inherently random about such a variable $X$, and all of what we consider to be "random" is described by the probability measure $P$, which assignes to each sample or set of samples $\omega\in\Omega$ some probability of occurring. The random variable $X$ then allows this probability to exist on our space $\mathbb{C}^n$ by assigning the probability it would assign to $\omega$ but instead to $X(\omega)$. 

Our probabilities are represented by projection operators $P$, and we will consider these rather than states in evolution through the Lindblad equation.



Through the inhabitance of random variables $X$ on the Hilbert space $\mathbb{C}^n$ (equipped with the standard Hermitian inner product), we may examine how symmetries in the system impact the behavior of "random" events. In particular, symmetries are not all spatially geometric (although they can be), but they impact the jump/hopping/transition probabilities through the jump/hopping terms in the transition matrix, which describes how likely states are to transition to other states in our quantum system. 

The transcendent part here is that symmetries affect the matrix terms and therefore describe what can and cannot happen in a system. 

Thus, we can understand "decoherence free subspaces"

## SDEs

In the Lindbladian, we see a factor of $\sqrt{dt}$ in the Kraus-operator derivation. In the Kraus-operator derivation, one writes short-time measurement/noise operators like
$$
M_0 = \mathbb{I}-\left(i H +\frac12\sum_a L_a^*L_a\right)dt
$$
and 
$$
M_a = \sqrt{dt} L_a. 
$$
Then,
$$
M_a\rho M_a^* = dt L_a\rho L_a^*,
$$
which is why $\sqrt{dt}$ appears. It ensures that the probability of a jump is $O(dt)$ and not $O(\sqrt{dt})$. This is exactly analogous to Ito calculus, where
$$
dW_t\sim\sqrt{dt},\quad (dW_t)^2 = dt
$$
But the Lindblad equation itself is the averaged deterministic equation, while the stochastic object is an unraveling: quantum jumps, quantum state diffusion, or stochastic master equations under continuous measurement.

\frac{d}{dt}\rho = -i[\frac{\Omega_R}{2}\sigma_x, \rho] + \gamma\left(\sigma_{-}\rho\sigma_{+} - \frac12\{\sigma_{+}\sigma_{-},\rho\}\right)

$$


### COntinious normalizing Flow

A continuous normlalizing flow defines a deterministic flow
$$
\frac{d}{dt}\phi_t(x) = v_t(\phi_t(x)),\quad \phi_0=\mathbb{I}
$$
A CNF constructs a time-dependent diffeomorphic map by solving an ODE, and pushes forward a base density as
$$
p_t=[\phi_t]_\sharp p_0.
$$
ALSO, it gives the instantanoeus change of variables formula involving the divergence of the vector field.
The density $p_t$ therefore satisfies the continuity/Liouville equation
$$
\partial_t p_t(x)=-\nabla\cdot (p_t(x)v_t(x)).
$$
Thus, **A CNF is a deterministic transport equationon probability densities** OR, equivalently, it is **A Markov process with transition kernel**
$$
K_t(x, dy)=delta_{\phi_t(x)}(dy).
$$
Its **generator** on test functions is 
$$
\mathcal{L}_t f = v_t\cdot\nabla f = \langle v_t, \nabla f\rangle.
$$
As such, its **adjoint on densities** is
$$
\mathcal{L}_t^* p = -\nabla\cdot (p v_t).
$$

### MArkov Kernel

A Markov Kernel is a positive mass-preserving operator:
$$
K 1 = 1, \quad K f\geq 0 \text{if} f\geq 0.
$$
The **collection** of Markov Kernels forms a **semigroup**, not a group. Most are not invertible as Markov Kernels - they smooth mix, reject, collapse, or forget information. That is the opposite of unitary reversibility.

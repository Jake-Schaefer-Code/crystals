```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def invtanh(t):
	return 0.5 * np.log((1+t)/(1-t))

def tanhe(t):
	return (-1+np.exp(2*t))/(1+np.exp(2*t))

def tanhT(t, r):
	return 0.5 * (tanhe(t+r)+tanhe(t-r))

def tanhR(t, r):
	return 0.5 * (tanhe(t+r)-tanhe(t-r))


def invarctan(t):
	return np.tan(np.pi/4*t)

def arctanT(t,r):
	return 1/np.pi*(np.arctan(t+r)+np.arctan(t-r))

def arctanR(t,r):
	return 1/np.pi*(np.arctan(t+r)-np.arctan(t-r))
```

```python
Nlines = 6
X,Y = np.meshgrid(*[np.linspace(0.001,2, 999), 1/(Nlines+1) * np.linspace(-Nlines, Nlines, 2*Nlines+1)])
U = arctanT(invarctan(X), invarctan(2*Y))
V = arctanR(invarctan(X), invarctan(2*Y))

fig = plt.figure(figsize=(10,10))
ax = fig.add_subplot(111)

# for i in range(U.shape[1]):
#     ax.plot(U[:,i], V[:,i], c='black')

for j in range(U.shape[0]):
    ax.plot(U[j,:], V[j,:], c='black')

ax.set_aspect('equal')

# plt.scatter(U,V)

plt.show()
```

```python
Nlines = 6
line_width = 2
X,Y = np.meshgrid(*[np.linspace(0.001,1,99), 1/(Nlines+1) * np.linspace(-Nlines, Nlines, 2*Nlines+1)])
U = tanhT(invtanh(X), 2*invtanh(Y))
V = tanhR(invtanh(X), 2*invtanh(Y))

fig = plt.figure(figsize=(10,10))
ax = fig.add_subplot(111)

# for i in range(U.shape[1]):
#     ax.plot(U[:,i], V[:,i], c='black')

for j in range(U.shape[0]):
    ax.plot(U[j,:], V[j,:], c='black', linewidth=line_width)
    ax.plot(-U[j,:], -V[j,:], c='black', linewidth=line_width)

V = tanhT(invtanh(X), 2*invtanh(Y))
U = tanhR(invtanh(X), 2*invtanh(Y))

for j in range(U.shape[0]):
    ax.plot(U[j,:], V[j,:], c='black', linewidth=line_width)
    ax.plot(-U[j,:], -V[j,:], c='black', linewidth=line_width)
    
ax.set_aspect('equal')

# plt.scatter(U,V)

plt.show()
```

MMC tends to 
$$
D(p_0\|\pi),
$$
so the relative entropy of the initial distribution and the stationary distribution. This is the measure of inefficiency of assuming (when are we assuming? from what?) the distribution is $\pi$ when the true dist is $p_0$.

$(n-1)$-simplex:
$$
\Delta^{n-1} = \left\{p\in\mathbb{R}^n \, : \, p_i\geq 0, \, \sum_{i=1}^n p_i = 1 \right\}
$$
has $i$-th face:
$$
F_i = \{p: p_i=0\},
$$
and the coordinate $p_i$ is, up to a constant factor, the perpendicular distance from face $i$.

For the standard simplex embedded in $\mathbb{R}^n$, 
$$
d(p, F_i) = \sqrt{\frac{n}{n-1}}p_i
$$

For uniform distribution, 
$$
D(u\|p) = - \log n - \frac{1}{n}\sum_i \log p_i,
$$
but note that 
$$
\frac{1}{n}\sum_i \log p_i = \log\left[\left(\prod_i p_i\right)^{1/n}\right] = \log[\mathrm{GM}(p)],
$$
which is the logarithm of the geometric mean, so 
$$
D(u\|p) = -\log [n\mathrm{GM}(p)].
$$
Also, 
$$
D(u\|p) = -\log \left[n\sqrt{\frac{n-1}{n}}\left(\prod_i d(p, F_i)\right)^{1/n}\right] = -\log \left[\sqrt{n(n-1)}\left(\prod_i d(p, F_i)\right)^{1/n}\right]
$$

So geometrically,

D(u∥p) measures the negative log of the multiplicative clearance of p from all faces of the simplex.

It is not a Euclidean distance from the center. Two points at the same Euclidean distance from the centroid can have different D(u∥p), because KL is extremely sensitive to approaching a face.

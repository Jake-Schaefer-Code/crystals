If we have nonlinearly temporally varying rate matrix $K(t)$, then the instantaneous MMC, 
$$
\begin{equation}
  -\frac{d}{dt} D(e^{K(t)} p||e^{K(t)} q)
\end{equation}
$$
is different at every instant in time. Therefore, choosing a single prior $q$ for the integral doesnt make sense:
$$
\begin{equation}
  \int_{[0,T]} D_K(p_t\| q) dt = \int_{[0,T]} -\frac{d}{ds} D(e^{K(s)} p_t||e^{K(s)} q) dt
\end{equation}
$$

 (see equation (I2) and (I4), where $q$ was the same throughout the whole integral because $K$ was the same). 

Thus, over an interval [0, T], the total mismatch cost should be described as 
\int_{[0,T]} D_K(p_t\| q_t) dt,
i.e., with time-varying q_t.

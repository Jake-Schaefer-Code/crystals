### Thermodynamics

- typically confined to/likes to describe equilibrium states of macroscopic systems or the transitions between such states


Thermodynamics can be extended to cover the nonequilibrium dynamics of small-scale systems.

### Nonequilibrium Thermodynamics

- on one side, can be seen as including microscopic dynamical properties into the far from equilibrium irreversible realm
- on the other side, they have led to the reassessment of the cornerstone of thermodynamics, namely the second law of thermodynamics, which is replaced by a much deeper symmetry relation, embodied in the integral and detailed fluctuation theorems
  - thus, the traditionally equilibrium notion can be extended to nonequilibrium by exploiting symmetries

- can address new questions:
  - eﬃciency at maximum power
  - information to work conversion
- Or, relating to the thermodynamic descriptions of small systems, e.g., the discussion of Brownian motors and refrigerators, efficiency of quantum dots and other small scale devices

### grand canonical equilibrium distribution
$$
P_m^{eq} = \exp\{-\beta(\epsilon_m -\mu n_m -\Omega^{eq})\} = \exp\{-\beta(\epsilon_m -\mu n_m)\}\exp\{\beta\Omega^{eq}\}
$$


The (equilibrium) grand potential $\Omega^{eq}$ follows from the normalization of the grand canonical equilibrium distribution $P_m^{eq}$:
$$
\exp\{-\beta\Omega^{eq}\} = \sum_m \exp\{-\beta(\epsilon_m -\mu n_m)\}
$$


- Why is $\Omega^{eq}$ a potential? why grand? why exponentiated? why the exponent of the normalization constant

### Equilibrium

The crucial property that is required from the rates at equilibrium is that of **detailed balance**:
$$
K_{m,m'}P^{eq}_{m'} = K_{m',m}P^{eq}_m,
$$
and combining this with the expression for the equilibrium distribution $P^{eq}_m$, we get that 
$$
1 = \dfrac{K_{m,m'}P^{eq}_{m'}}{K_{m',m}P^{eq}_m} 
= \dfrac{K_{m,m'}\exp\{-\beta(\epsilon_m -\mu n_m)\}}{K_{m',m}\exp\{-\beta(\epsilon_{m'} -\mu n_{m'})\}}
= \dfrac{K_{m,m'}}{K_{m',m}}\exp\{-\beta(\epsilon_m - \epsilon_{m'} - \mu n_m + \mu n_{m'})\},
$$
so, from $\beta=\frac{1}{k_B T}$,
<!-- $$ 
\ln\dfrac{K_{m',m}}{K_{m,m'}} = \beta(\epsilon_m - \epsilon_{m'} - \mu (n_m - n_{m'}))
$$ -->
<!-- and thus, from $\beta=\frac{1}{k_B T}$, -->
$$
k_B\ln\dfrac{K_{m',m}}{K_{m,m'}} = \dfrac{\epsilon_m - \epsilon_{m'} - \mu (n_m - n_{m'})}{T} = \dfrac{q_{m,m'}}{T},
$$
where $q_{m,m'}$ is the "elementary" heat absorbed by the system to make the transition $m'\to m$.

**NOTE:** in the presence of driving, which is shifting the energy levels in time, this relation is supposed to hold *at each moment in time*. Thus, as $\epsilon_m(t)$ is a function of time, the rates $K_{m,m'}(t)$ also become time-dependent.


**state functions:** quantities that depend on probability distribution $P_m$ of the system, but *not* on the way this distribution was achieved.

### Nonequilibrium

A single reservoir does not exclude nonequilibrium altogether. Time-dependent driving can still force the system out of equilibrium and can produce a periodically driven nonequilibrium state.

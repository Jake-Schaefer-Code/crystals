## Operations on Representations
Classifying the representations of a group, up to isomorphism, amounts to classifying collections of matrices satisfying the group’s relations up to simultaneous conjugacy. Having an invariant, i.e., a way of attaching an algebraic object to each square matrix that can detect if a pair of matrices are not conjugate, would be invaluable in this endeavor. So: what conjugacy invariants of square matrices are there? As we have seen, spectral theory plays a special role in studying representations—a natural choice of invariant, then, might be associating a matrix $A$ to its set of eigenvalues. Remember that the eigenvalues
are themselves determined by its characteristic polynomial
$$
p_A(t) = t^n + a_{n-1}t^{n-1}+\cdots+a_1 t+ a_0,
$$
another conjugacy invariant. Among the coefficients are
$$
a_{n-1}= -(\lambda_1+\cdots+\lambda_n) = -\mathrm{Tr}A, \\
a_0 = (-1)^n \lambda_1\cdots\lambda_n = (-1)^n\mathrm{det}A,
$$
where the $\lambda_i$ are the solutions of $p_A(t)$ recorded with multiplicity. A fundamental object of study in the sequel is the character of a representation $\rho$, i.e., the function sending elements $g \in G$ to the trace of the transformation $\rho(g)$. In some sense, this is because knowing the trace of a matrix $A$ and its powers is equivalent to knowing its characteristic polynomial. Indeed, since 
$Ax=\lambda x$ means that 
$A^k x = \lambda^k x$, we always have
$$
\mathrm{Tr}(A^k) = \lambda_1^k+\cdots+\lambda_n^k
$$
Hence the generating function1 for the sequence
$$
\mathrm{Tr}(\mathbb{I}), \mathrm{Tr}(A), \mathrm{Tr}(A^2), \dots , \mathrm{Tr}(A^k), \dots
$$
is given by
$$
\sum_{k=0}^\infty t^k\mathrm{Tr}(A^k)
= \sum_{k=0}^\infty\left(\sum_{i=1}^n\lambda_i^k\right)t^k 
= \sum_{i=1}^n\left(\sum_{k=0}^\infty(\lambda_i t)^k\right)
= \sum_{i=1}^n\frac{1}{1-\lambda_i t}
$$
This formula allows us to recover the set of eigenvalues of $A$ from only the traces of its powers!

Assume representations are finite-dimensional complex representations.

Let

$$\pi:G\twoheadrightarrow G/N
$$
be the quotient map. We want to build irreducible representations of (G) whose kernels cut out exactly (N).

Set

$$
H=G/N.
$$
The main fact we need is:

$$\bigcap_{\sigma \in \operatorname{Irr}(H)} \ker \sigma={e_H}.
$$
Why? Because the regular representation of (H),

$$\lambda:H\to GL(\mathbb C[H]),
\qquad
\lambda(h)e_x=e_{hx},
$$
is faithful. Indeed, if (h\neq e_H), then

$$\lambda(h)e_{e_H}=e_h\neq e_{e_H},
$$
so (\lambda(h)\neq I). Hence

$$\ker \lambda={e_H}.
$$
By Maschke’s theorem, the regular representation decomposes as a direct sum of irreducible representations:

$$\mathbb C[H]\cong \bigoplus_{\sigma\in \operatorname{Irr}(H)} V_\sigma^{\oplus m_\sigma}.
$$
The kernel of a direct sum representation is the intersection of the kernels of its summands, so

$$\{e_H\}= \ker \lambda=\bigcap_{\sigma\in \operatorname{Irr}(H)} \ker \sigma.
$$
Now lift each irreducible representation of (H=G/N) to (G). For every

$$\sigma:H\to GL(V_\sigma),
$$
define

$$\widetilde{\sigma}:G\to GL(V_\sigma)
$$
by

$$\widetilde{\sigma}(g)=\sigma(gN).
$$
By part (a), (\sigma) is irreducible as a representation of (G/N) if and only if (\widetilde{\sigma}) is irreducible as a representation of (G). Therefore the lifted representations (\widetilde{\sigma}) are irreducible (G)-representations.

Now compute the intersection of their kernels:

$$\bigcap_{\sigma\in \operatorname{Irr}(G/N)} \ker \widetilde{\sigma}
=

\bigcap_{\sigma\in \operatorname{Irr}(G/N)} \ker(\sigma\circ \pi).
$$
But

$$
g\in \ker(\sigma\circ \pi)
\iff
\sigma(\pi(g))=I.
$$
So

$$
g\in \bigcap_{\sigma\in \operatorname{Irr}(G/N)} \ker(\sigma\circ \pi)
\iff
\pi(g)\in \bigcap_{\sigma\in \operatorname{Irr}(G/N)} \ker \sigma.
$$
The intersection on the right is trivial, so

$$\pi(g)\in {e_{G/N}}.
$$
Therefore

$$
g\in \ker \pi=N.
$$
Hence

$$\bigcap_{\sigma\in \operatorname{Irr}(G/N)} \ker \widetilde{\sigma}
=

N.
$$
So if (S) is the subset of irreducible (G)-representations obtained by lifting the irreducible representations of (G/N), then

$$\boxed{
N=\bigcap_{V\in S}\ker \rho_V.
}
$$
This proves the claim.

The character table detects this because for an irreducible representation (\rho_V) with character (\chi_V),

$$
g\in \ker \rho_V
\iff
\rho_V(g)=I
\iff
\chi_V(g)=\chi_V(e)=\dim V.
$$
Thus the kernel of each irreducible representation can be read from its character row, and normal subgroups arise as intersections of such kernels.

```python
def latex_bmatrix(A, fmt=".3g"):
  A = onp.asarray(A)
  rows = [" & ".join(format(x, fmt) for x in row) for row in A]
  body = r" \\ ".join(rows)
  return rf"\begin{{bmatrix}}{body}\end{{bmatrix}}"

pmath = lambda data: display(Math(data))

def K_matrix_to_sympy(MK):
  return sp.Matrix([[K.to_sympy(x) for x in row] for row in MK])

```

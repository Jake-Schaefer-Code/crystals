$$
\Delta x = v_0 t + \frac12 at^2
$$

```python
import numpy as onp
import matplotlib.pyplot as plt
```

```python
ts = onp.linspace(0,1,100)
v_0 = 0.5 * 9.81
a = -9.81
y_0 = 1.0

Δy = v_0 * ts + 0.5 * a * ts**2
qt = onp.random.normal(0, 0.05, size=ts.shape)
f = 5
pt = 0.05 * onp.sin(2 * onp.pi * f * ts)
```

plotting perturbations $q(t)$ and $p(t)$

```python
plt.plot(ts, y_0 + Δy, c='blue')
plt.plot(ts, y_0 + Δy + qt, c='red')
plt.plot(ts, y_0 + Δy + pt, c='green')
plt.show();
```

```python
import numpy as np
from fractions import Fraction
np.set_printoptions(edgeitems=20, linewidth=100000)

import sympy as sp
sp.init_printing(use_unicode=True)

import matplotlib.pyplot as plt
```

```python
rho = np.array([[1/4,1/4,0],[1/4,1/4,0],[0,0,1/2]])
sigma = np.array([[1/2,0,0],[0,1/4,1/4],[0,1/4,1/4]])
A = 1/6 * rho - 5/6 * sigma
print(A)
print(np.linalg.eig(A)[0])
print(np.sum(np.abs(np.linalg.eig(A)[0])))

print(np.sum(np.abs(np.linalg.eig(A @ (-1/6*sigma+np.eye(3)))[0])))
```

```python
n = 1000

K=np.array([[1-2*k/n-2*l/n for k in range(n)] for l in range(n)])
print(K)

x_vals = np.array([[i/n for i in range(n)] for j in range(n)])
print(K.shape, x_vals.shape)

output = np.zeros((n,n,2))
output[:,:,0] = x_vals
output[:,:,1] = K
print(output)

ev = []
ev2 = []
for i in range(1, 100):
    n1 = 10*i
    K=np.array([[1-2*k/n1-2*l/n1 for k in range(n1)] for l in range(n1)])
    ev.append(np.linalg.eig(K)[0][0])
    ev2.append(np.linalg.eig(K)[0][1])

print(ev)
plt.plot([i*10 for i in range(1, 100)], ev)
plt.plot([i*10 for i in range(1, 100)], ev2)
plt.legend()
plt.show()

plt.plot(output[0][:,0], output[0][:,1])
plt.plot(output[1][:,0], output[1][:,1])
plt.show()
```

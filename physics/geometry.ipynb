In numerical relativity, gauge conditions are specific choices of coordinate systems (or gauges) that simplify the equations of general relativity and help maintain numerical stability during simulations. These conditions determine how the spacetime coordinates evolve during the simulation, which can significantly impact the accuracy and stability of the numerical solution.

Two critical gauge conditions in numerical relativity are:

Lapse Function (α): Determines the rate of progression of time.

Shift Vector (β^i): Determines how the spatial coordinates shift relative to the coordinate grid.
